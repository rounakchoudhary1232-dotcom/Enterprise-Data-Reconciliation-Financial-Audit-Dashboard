```python
query = """
SELECT u.*
FROM users u
JOIN (
    SELECT user_id, MAX(updated_at) AS latest_time
    FROM users
    GROUP BY user_id
) latest
ON u.user_id = latest.user_id
AND u.updated_at = latest.latest_time
WHERE u.status = 'Active';
"""

active_users = pd.read_sql(query, conn)

active_users
```

```python
import pandas as pd
import re

logs = pd.read_csv("server_logs.txt")

pattern = r'\[(.*?)\].*?INFO.*?"u":"(.*?)".*?"item_code":"(.*?)".*?"eur_val":([0-9.]+)'

data = []

for line in logs["Raw_Server_Log"]:
    match = re.search(pattern, str(line))
    if match:
        data.append(match.groups())

logs_df = pd.DataFrame(
    data,
    columns=["date", "user_id", "product_id", "eur_value"]
)

logs_df.head()
```

```python
rates = pd.read_csv("daily_exchange_rates.csv")

rates.head()
```

```python
print(rates.columns)
```

```python
logs_df["date"] = pd.to_datetime(logs_df["date"]).dt.date
rates["Date"] = pd.to_datetime(rates["Date"]).dt.date
```

```python
merged_df = pd.merge(
    logs_df,
    rates,
    left_on="date",
    right_on="Date",
    how="left"
)

merged_df.head()
```

```python
final_df = merged_df[
    [
        "date",
        "user_id",
        "product_id",
        "eur_value",
        "EUR_to_USD",
        "usd_value"
    ]
]

final_df.head()
```
